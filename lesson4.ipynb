```python
fig,ax = plt.subplots()
sotuv1 = [1,6,7,8]
oylik1 = [3,5,7,6]

sotuv2 = [3,2,4,5]
oylik2 = [2,1,3,6]

ax.plot(sotuv1,oylik1,
        color = "red",
        linestyle =":",
        linewidth = 2,
        marker = "o",
        label="A mahsulot",) #chiziq + legenda nomi
ax.plot(sotuv2, oylik2,
        label ="B mahsulot",
        color = "blue",
        linestyle ="--",
        linewidth = 2,
        marker = "o",)

ax.set_title("Oylik sotuv") #grafik sarlavhasi - tepada
ax.set_xlabel("Oy") #X o'qi nomi -pastda
ax.set_ylabel("Sotuv (mln)") #Y o'qi nomi - chapda
ax.set_facecolor("#c3d9c8")

ax.legend() #har bir chiziq nomi - burchakda
ax.grid(True) #fon chiziqlari - o'qishni osonlashtiradi
#professional grafikda odatda faqat chap va past qoldiriladi
plt.show()
```

```python
shaharlar = ["Toshkent", "Buxoro", "Samarqand", "Namangan"]
sotuv = [200, 300, 100, 500]

fig, ax = plt.subplots()
ax.bar(shaharlar, sotuv,
       width = 0.8,
       color = "#fce6f2",
       edgecolor = "#fe3991",)
ax.set_title("Shaharlar bo'yicha sotuv")
ax.set_xlabel("Shahar")
ax.set_ylabel("Sotuv (mln)")
ax.grid(axis="y")
plt.show()
```

```python
kurslar = ["Python asoslari",
            "Machine Learning",
            "Deep Learning",
            "Data Science"]
sotuv = [120, 95, 60, 85]

fig, ax = plt.subplots()
ax.barh(kurslar, sotuv,
        height = 0.6,
        color = "#d5eb7a",
        edgecolor = "#618505",
)

ax.set_title("Kurslar bo'yicha talabalar soni")
ax.set_xlabel("Talabalar soni")
ax.grid(axis="x")
plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(26,13), dpi = 120, facecolor = 'white')

shaharlar = ["Toshkent", "Buxoro", "Samarqand", "Namangan"]
sotuv = [200, 300, 100, 500]

kurslar = ["Python asoslari", "Machine Learning",
            "Deep Learning",
            "Data Science"]
sotuv = [120, 95, 60, 85]

ax[0][0].bar(shaharlar, sotuv,
       width = 0.6,
       color = "steelblue",
       edgecolor = "black",)
ax[0][0].set_title("Shaharlar bo'yicha sotuv")
ax[0][0].set_xlabel("Shahar")
ax[0][0].set_ylabel("Sotuv (mln)")
ax[0][0].grid(axis="y")
ax[0][0].set_facecolor("#dbdbdb")

ax[1][0].barh(kurslar, sotuv,
        height = 0.6,
        color = "green",
        edgecolor = "black",
)

ax[1][0].set_title("Kurslar bo'yicha talabalar soni")
ax[1][0].set_xlabel("Talabalar soni")
ax[1][0].grid(axis="x")
ax[1][0].set_facecolor("#dbdbdb")

ax[0][1].plot([4,5,6], [-1,-2,-3])
ax[0][1].set_title("Mening grafikim 3")
ax[0][1].set_xlabel("X o'qi i")
ax[0][1].set_ylabel("Y o'qi nomi")
ax[0][1].set_facecolor("#dbdbdb")

ax[1][1].plot([1,2,3], [4,5,6])
ax[1][1].set_title("Mening grafikim 4")
ax[1][1].set_xlabel("X o'qi i")
ax[1][1].set_ylabel("Y o'qi nomi")
ax[1][1].set_facecolor("#dbdbdb")

plt.show()

```

```python
import numpy as np
x = np.random.randint(40, 200, 80)
y = 3.8 * x  + np.random.normal(0, 30, 80)

fig, ax = plt.subplots()
ax.scatter(x, y,
           s = 40,
           c = "steelblue",
           alpha = 0.8,
           )
ax.set_title("Uy maydoni vs Narxi")
ax.set_xlabel("Maydon (m²)")
ax.set_ylabel("Narx (mln so'm)")
ax.grid(True)

plt.show()
```

```python
fig, ax = plt.subplots()
ulushlar = [40, 30, 20, 10]
nomlar = ["Python", "SQL", "ML", "Boshqa"]

ax.pie(ulushlar,
       labels = nomlar,
       autopct="%1.2f%%",
       explode = [0,0,0,0.1]
       )
ax.set_title("Kurs to'lovi")

plt.show()
```
